# Step 04: Model Fine-Tuning & Checkpointing

This notebook trains/fine-tunes the Piper Arabic model using **PyTorch Lightning** and **VITS** architecture. Checkpoints and TensorBoard logs are saved directly to Google Drive for automatic resumption.

### Pipeline Steps:
1. Detect latest checkpoint in Drive to enable automatic resume
2. Launch TensorBoard visualization inline
3. Execute fine-tuning process with `piper_train`

```python
# 1. Checkpoint Detection & Setup
import os
from pathlib import Path

ckpt_dir = Path('/content/drive/MyDrive/Arabic-Piper/checkpoints/experiment001')
ckpt_dir.mkdir(parents=True, exist_ok=True)

existing_ckpts = sorted(list(ckpt_dir.glob('*.ckpt')))
if existing_ckpts:
    resume_ckpt = str(existing_ckpts[-1])
    print(f"Resuming training from latest checkpoint: {resume_ckpt}")
else:
    base_ckpt = Path('/content/drive/MyDrive/Arabic-Piper/checkpoints/base/ar_JO-kareem-medium.ckpt')
    resume_ckpt = str(base_ckpt) if base_ckpt.exists() else ''
    print(f"Starting fine-tuning from base checkpoint: {resume_ckpt}")
```

```python
# 2. Launch TensorBoard
%load_ext tensorboard
%tensorboard --logdir /content/drive/MyDrive/Arabic-Piper/tensorboard/experiment001
```

```python
# 3. Execute Training Run
!python -m piper_train \
    --dataset-dir /content/drive/MyDrive/Arabic-Piper/processed/experiment001 \
    --accelerator gpu \
    --devices 1 \
    --batch-size 16 \
    --validation-split 0.05 \
    --max-epochs 50 \
    --checkpoint-epochs 5 \
    --default_root_dir /content/drive/MyDrive/Arabic-Piper/checkpoints/experiment001 \
    --resume_from_checkpoint "{resume_ckpt}"
```
